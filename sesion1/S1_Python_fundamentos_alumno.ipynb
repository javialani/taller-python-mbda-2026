# 2. Variables y tipos de datos

Una **variable** es un **tarro con etiqueta**: la etiqueta es el nombre y dentro guardamos un valor.

- **Guardar:** `edad = 22`. El signo `=` no significa "es igual": significa **guarda**.
- **Usar:** al escribir el nombre, Python saca el valor para usarlo; el tarro no se vacía.
- **Cambiar:** si vuelves a usar `=`, el valor anterior se tira y entra el nuevo. Un tarro guarda **un solo valor**.

```python
edad = 22          # guardar
print(edad + 1)    # usar: 23 (edad sigue valiendo 22)
print(edad)

edad = 23          # cambiar: el 22 se sustituye por 23
print(edad)
```

**Nombres de variables:** sin espacios (`tarifa_media`, no `tarifa media`), sin empezar por número (`edad2` sí, `2edad` no) y, por costumbre, en minúscula con guiones bajos. Python distingue mayúsculas: `Edad` y `edad` son **tarros distintos**.

### Tipos de datos
Cada valor tiene un **tipo**, y el tipo decide qué puedes hacer con él (sumar números, pasar un texto a mayúsculas…):

| Tipo | Qué es | Ejemplo Titanic |
|---|---|---|
| `int` | número entero | clase = 3 |
| `float` | número decimal | tarifa = 7.25 |
| `str` | texto | nombre = "Braund, Mr. Owen Harris" |
| `bool` | verdadero/falso | sobrevivio = False |
| `None` | "no hay dato" | edad desconocida |

```python
nombre = "Braund, Mr. Owen Harris"
clase = 3
edad = 22
tarifa = 7.25
sobrevivio = False

print(type(nombre), type(clase), type(tarifa), type(sobrevivio))
```

```python
# Cada tipo tiene sus operaciones (en la §5 las veremos todas)
print(tarifa * 2)          # con números: multiplicar
print(edad + 1)            # con números: sumar
print(nombre.upper())      # con textos: pasar a mayúsculas
```

**f-strings**: la forma cómoda de construir frases con datos. Pon una `f` delante y las variables entre `{}`.

```python
print(f"{nombre} viajaba en {clase}ª clase y pagó {tarifa} libras.")
print(f"Tarifa con 2 decimales: {tarifa:.2f}")
```

### Cuidado con los tipos
Los datos que leemos de un fichero suelen llegar **como texto**. `"22" + 1` falla; hay que convertir.

```python
edad_texto = "22"
edad_num = int(edad_texto)     # también float(), str(), bool()
print(edad_num + 1)
```

# 5. Operadores: calcular, comparar y combinar

Tres familias de operadores que usaremos en todo el taller:
1. **Aritméticos**: Python como calculadora.
2. **De comparación**: preguntas de sí o no (`True` / `False`).
3. **Lógicos**: combinar varias preguntas.

### 5.1 Aritméticos

| Operador | Qué hace | Ejemplo |
|---|---|---|
| `+` `-` | suma, resta | `1912 - edad` |
| `*` | multiplicación | `tarifa * 1.10` |
| `/` | división (siempre da decimales) | `71.28 / 7.25` |
| `//` | división entera | `891 // 12` → 74 |
| `%` | resto (lo que sobra al repartir) | `891 % 12` → 3 |
| `**` | potencia | `1.05 ** 3` |

Orden: primero `**`, luego `*` `/` `//` `%`, y al final `+` `-`. Los paréntesis mandan.

```python
print(edad + 1)                   # 23 → edad el próximo año
print(1912 - edad)                # 1890 → año de nacimiento aproximado
print(round(tarifa * 1.10, 2))    # subir la prima un 10 %
print(round(71.28 / 7.25, 2))     # / siempre da decimales
print(891 // 12, 891 % 12)        # 74 lotes completos de 12 expedientes… y sobran 3
print(round(100 * 1.05 ** 3, 2))  # prima de 100 que sube un 5 % cada año, 3 años
```

**¿Para qué sirve el resto `%`?** Para saber si un número es par (`n % 2 == 0`), para hacer algo cada N filas (revisar uno de cada 10 expedientes: `i % 10 == 0`) o para repartir en turnos.

# 6. Decidir y repetir: `if` y `for`

- `if` = **tomar decisiones** ("si sobrevivió, cuéntalo").
- `for` = **repetir** una acción para cada elemento ("para cada pasajero…").

⚠️ En Python la **sangría** (los 4 espacios) no es estética: indica qué está "dentro" del `if` o del `for`.

```python
for p in mini_titanic:
    if p["sobrevivio"] == 1:
        estado = "✅ sobrevivió"
    else:
        estado = "❌ no sobrevivió"
    print(f'{p["nombre"]:<35} {estado}')
```

```python
# Patrón "contador": empieza en 0 y suma cuando se cumple la condición
supervivientes = 0
for p in mini_titanic:
    if p["sobrevivio"] == 1:
        supervivientes += 1

tasa = supervivientes / len(mini_titanic)
print(f"Sobrevivieron {supervivientes} de {len(mini_titanic)} ({tasa:.0%})")
```

En la condición del `if` puedes usar todos los operadores de la §5: `==`, `!=`, `<`, `and`, `or`, `not`… y también `is None`.

# 7. Funciones: recetas reutilizables

Llevamos un rato copiando el mismo bucle de "contar supervivientes". Una **función** es una **receta**: la escribes una vez, le das ingredientes (**parámetros**) y te devuelve un resultado (`return`).
Ya has usado muchas: `print`, `len`, `sum`…

```python
def tasa_supervivencia(pasajeros):
    """Devuelve la proporción de supervivientes (0 a 1)."""
    vivos = 0
    for p in pasajeros:
        vivos += p["sobrevivio"]
    return vivos / len(pasajeros)

print(f"{tasa_supervivencia(mini_titanic):.0%}")
```

```python
# Combinada con una comprehension, se lee casi como una frase
mujeres = [p for p in mini_titanic if p["sexo"] == "female"]
hombres = [p for p in mini_titanic if p["sexo"] == "male"]
print(f"Mujeres: {tasa_supervivencia(mujeres):.0%} | Hombres: {tasa_supervivencia(hombres):.0%}")
```

# 8. NumPy: la calculadora industrial

Python "puro" va elemento a elemento, como **una carretilla**. **NumPy** trabaja con columnas enteras de golpe, como **una cinta transportadora**: es la base de Pandas, scikit-learn y casi todo el análisis de datos en Python.

Concepto clave: el **array**, una lista en la que **todos los elementos son del mismo tipo** (normalmente números).

```python
import numpy as np   # "np" es el alias universal

tarifas = np.array([p["tarifa"] for p in mini_titanic])
clases = np.array([p["clase"] for p in mini_titanic])
vivos = np.array([p["sobrevivio"] for p in mini_titanic])

print(tarifas)
print(tarifas.dtype, tarifas.shape)
```

### Vectorización: una operación, toda la columna
Sin bucles: es como aplicar un descuento a todo el catálogo de golpe, en lugar de cambiar los precios etiqueta a etiqueta.

```python
# Con una lista necesitarías un bucle; con NumPy, no.
print((tarifas * 1.10).round(2))   # subida del 10 % a las 12 tarifas a la vez
print(tarifas > 20)                # ¡comparar también es vectorial!

# Todos los elementos de un array son del mismo tipo: si mezclas, todo se convierte en texto
print(np.array([22, "Braund"]))
```

```python
# Estadísticos en una línea. Fíjate: la media es mucho mayor que la mediana,
# porque unos pocos billetes de primera clase muy caros tiran de ella.
print(f"Media: {tarifas.mean():.2f} | Mediana: {np.median(tarifas):.2f}")
print(f"Desv. típica: {tarifas.std():.2f} | Mín: {tarifas.min()} | Máx: {tarifas.max()}")
print(f"Percentil 75: {np.percentile(tarifas, 75):.2f}")
```

### Máscaras booleanas: filtrar sin bucles
`tarifas > 20` devuelve un array de `True/False`. Si lo usas **dentro de los corchetes**, te quedas solo con los `True`. Funciona como un **colador**: solo pasan las filas que cumplen la condición.

```python
caros = tarifas > 20
print(tarifas[caros])

print("Tarifa media en 1ª clase:", tarifas[clases == 1].mean().round(2))
print("Tasa de supervivencia:", vivos.mean())   # ¡la media de 0/1 es una proporción!
```

### Datos que faltan: `np.nan`
En NumPy, "no hay dato" se escribe `np.nan` (*Not a Number*). Contagia los cálculos, así que usamos las versiones `nan-`.

```python
edades = np.array([p["edad"] if p["edad"] is not None else np.nan for p in mini_titanic])
print(edades)
print("mean:", edades.mean())        # nan: un dato desconocido "contamina" el resultado
print("nanmean:", np.nanmean(edades).round(1))
print("¿Cuántos faltan?:", np.isnan(edades).sum())
```

```python
import csv, urllib.request

URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
with urllib.request.urlopen(URL) as resp:
    lineas = resp.read().decode("utf-8").splitlines()

titanic = list(csv.DictReader(lineas))      # ¡una lista de diccionarios, como el mini-Titanic!
print(len(titanic), "pasajeros")
print(titanic[0])
```

Fíjate en `titanic[0]`: **todo es texto** (`'Age': '22'`), y las edades que faltan son `''`. Para calcular cualquier cosa hay que convertir tipos y tratar vacíos, columna a columna:

```python
edades = np.array([float(p["Age"]) if p["Age"] != "" else np.nan for p in titanic])
vivos = np.array([int(p["Survived"]) for p in titanic])
sexo = np.array([p["Sex"] for p in titanic])
clase = np.array([int(p["Pclass"]) for p in titanic])

print(f"Edad media: {np.nanmean(edades):.1f} · Edades que faltan: {np.isnan(edades).sum()}")
print(f"Supervivencia mujeres: {vivos[sexo == 'female'].mean():.0%} · hombres: {vivos[sexo == 'male'].mean():.0%}")
for c in [1, 2, 3]:
    print(f"Clase {c}: {vivos[clase == c].mean():.0%}")
```

Funciona… pero piensa en lo que ha costado:
- Convertir **a mano** cada columna y cada vacío.
- Arrays **sueltos** que hay que mantener alineados (si ordenas uno, se rompe todo).
- Nada de nombres de columna en los resultados, ni tablas legibles, ni gráficos.

Ahora imagina 12 columnas, 1 millón de filas y 3 ficheros que cruzar. **No escala.**

### Avance de la sesión 2 · Pandas = tablas que se filtran, ordenan y resumen solas

```python
import pandas as pd

df = pd.read_csv(URL)
df.groupby(["Sex", "Pclass"])["Survived"].mean().unstack().round(2)
```

Dos líneas. Lo que hoy nos ha costado una sesión, **la semana que viene es una línea**… y entenderás qué pasa por debajo.
